# Imports

```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import numpy as np
```

# Open all files

```python
df_dict = {}
directory = 'processed_data/processed_aero_data'
for filename in os.listdir(directory):
    if filename.endswith('.csv'):
        filepath = os.path.join(directory, filename)
        df_dict[filename] = pd.read_csv(filepath)
```

# Attempting to calculate controller efficiency ($\eta_{c}$)

### Available Variables:
- $I_{motor}$

- $V_{controller}$

- $V_{tractive}$

- $V_{battery}$

- $I_{battery}$

Using these variables, we can find the efficiency of the motor controller at every instance through the equation: $$\frac{I_{motor} \times V_{tractive}}{I_{battery} \times V_{controller}} \times 100\% = \eta_c$$

```python
for name, df in df_dict.copy().items():
    df['motor power (w)'] = df['real torque']*(df['motor rpm']*((2*np.pi)/60))
    df['controller efficiency'] = df['motor power (w)']/df['pack power']
    df_dict[name] = df
    plt.plot(df['time'], df['pack power'])
    plt.plot(df['time'], df['motor power (w)'])
    plt.show()
```
